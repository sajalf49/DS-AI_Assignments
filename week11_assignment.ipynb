## Step 1 — Simulated Text Dataset
I created a small set of example sentences related to transaction activities. Some represent normal operations, while others describe fraud-like situations.

```text
Transaction successful
Unusual transaction detected
Payment completed successfully
Unauthorized charge reported
Account locked due to fraud
User requested refund
Large withdrawal approved
Multiple login attempts detected
Suspicious payment declined
Transfer completed successfully
```

I used this dataset to perform various NLP preprocessing steps.


## Step 2 — Text Preprocessing
I applied standard text preprocessing using **NLTK** and **Scikit-learn**.

### Steps performed:
1. Converted all text to lowercase.  
2. Tokenized text into individual words.  
3. Removed stopwords (common words such as “the”, “is”, “and”).  
4. Lemmatized words (reduced to their root form).

### Example:
| Original Sentence | After Preprocessing |
|--------------------|--------------------|
| "Unauthorized charge reported" | ['unauthorized', 'charge', 'report'] |
| "Transaction successful" | ['transaction', 'successful'] |
| "Account locked due to fraud" | ['account', 'lock', 'fraud'] |

After cleaning, I obtained a clear set of informative tokens that represent the essence of each sentence.


## Step 3 — TF-IDF Feature Extraction
Next, I used **TF-IDF (Term Frequency–Inverse Document Frequency)** to convert words into numerical features.
